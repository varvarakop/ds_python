1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

```python
x = sp.symbols('x')

f_x = 10 * x - 5
diff_f_x = sp.diff(f_x)

print(diff_f_x)
```

```python
x = np.linspace(0, 10)
f = 10 * x - 5

plt.figure(figsize=(10, 5))
plt.plot(x, f, label='f(x) = 10 * x - 5', color='blue')
plt.axhline(10, label="f'(x) = 10", color='green')

plt.title("Графік функції та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```

**ВИСНОВОК:**

При невеликому збільшенні кількості годин навчання поблизу будь-якої точки `x` оцінка змінюється зі швидкістю 10. Оскільки похідна стала, ця швидкість однакова для будь-якої кількості годин навчання, тому "будь-якої точки".

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
x = sp.symbols('x')

f_sin = sp.sin(x)
diff_f_sin = sp.diff(f_sin)

f_ln = sp.ln(x)
diff_f_ln = sp.diff(f_ln)

print(diff_f_sin)
diff_f_ln
```

```python
def diff_plot(f, d_f):

  x = sp.symbols('x')

  f = sp.lambdify(x, f, 'numpy')
  d_f = sp.lambdify(x, d_f, 'numpy')

  x_values = np.linspace(0, 100)

  plt.figure(figsize=(30,5))
  plt.plot(x_values, f(x_values), label='f(x)')
  plt.plot(x_values, d_f(x_values), label="f'(x)")

  plt.title("Графік функції та її похідної")
  plt.xlabel("x")
  plt.ylabel("y")
  plt.legend()
  plt.grid(True)
```

```python
diff_plot(f_sin, diff_f_sin)
```

```python
diff_plot(f_ln, diff_f_ln)
```
